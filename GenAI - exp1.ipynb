```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "Artificial Intelligence is transforming the world."

inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(**inputs, max_new_tokens=100)

translation = tokenizer.decode(
    outputs[0], skip_special_tokens=True
)

print("English:", text)
print("French:", translation)
```

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# Experiment 1: Sentiment Analysis using PyTorch
sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    framework="pt"
)

text1 = "I really enjoyed learning Artificial Intelligence."

print("Sentiment Analysis Result:")
print(sentiment(text1))

# Experiment 1: English to French Translation
model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text2 = "Artificial Intelligence is transforming the world."
inputs = tokenizer(text2, return_tensors="pt")

with torch.no_grad():
    outputs = model.generate(**inputs, max_new_tokens=100)

translation = tokenizer.decode(
    outputs[0], skip_special_tokens=True
)

print("\nLanguage Translation:")
print("English:", text2)
print("French:", translation)
```
