```python
# 3.  Sequential Chain 
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate

llm = OllamaLLM(model="llama3.2")
```

```python
summary_prompt = PromptTemplate(
    input_variables=["paragraph"],
    template="""
Summarize the following paragraph in a concise manner.

Paragraph:
{paragraph}
"""
)
```

```python
bullet_prompt = PromptTemplate(
    input_variables=["summary"],
    template="""
Convert the following summary into exactly 5 bullet points.

Summary:
{summary}
"""
)
```

```python
summary_chain = summary_prompt | llm
bullet_chain = bullet_prompt | llm
```

```python
paragraph = input("Enter a paragraph:\n")

summary = summary_chain.invoke({
    "paragraph": paragraph
})

bullet_points = bullet_chain.invoke({
    "summary": summary
})

print("\nSUMMARY:")
print(summary)

print("\n5 BULLET POINTS:")
print(bullet_points)
```

```python
# 5. Retrieval-Augmented Generation (RAG)
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
from langchain_community.document_loaders import TextLoader

file_path = r"C:\Users\HP\Downloads\AI_Notes.txt"

loader = TextLoader(file_path)
documents = loader.load()

print("Documents loaded:", len(documents))
print(documents[0].page_content)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

```python
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
vector_db = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS vector database created successfully.")
```

```python
question = input("Ask a question: ")

retrieved_docs = vector_db.similarity_search(
    question,
    k=3
)

print("\nRetrieved Information:\n")

for i, doc in enumerate(retrieved_docs, 1):
    print(f"--- Document {i} ---")
    print(doc.page_content)
```

```python
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model="llama3.2")
```

```python
context = "\n\n".join(
    doc.page_content for doc in retrieved_docs
)

prompt = f"""
Answer the question using ONLY the provided context.

Context:
{context}

Question:
{question}

If the answer cannot be found in the context, say:
"I could not find the answer in the provided document."
"""

answer = llm.invoke(prompt)

print("\nAnswer:")
print(answer)
```
