```python
from pydantic import BaseModel, Field
from typing import List
from langchain_core.output_parsers import PydanticOutputParser

# Define the expected output format
class StudentInfo(BaseModel):
    name: str = Field(description="Student name")
    branch: str = Field(description="Student branch")
    skills: List[str] = Field(description="Technical skills")

# Create the parser
parser = PydanticOutputParser(pydantic_object=StudentInfo)

# Example structured response (simulated LLM output)
sample_output = """
{
    "name": "Ananya",
    "branch": "Computer Science",
    "skills": ["Python", "SQL", "Machine Learning"]
}
"""

# Parse and validate the response
result = parser.parse(sample_output)

print("Parsed LLM Output:")
print(result.model_dump())
```
